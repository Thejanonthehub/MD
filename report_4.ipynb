Stress Strain Curve

![main-qimg-80178c75335ca14e994afa83867f94be-lq.jpeg](md_r4/main-qimg-80178c75335ca14e994afa83867f94be-lq.jpeg)

In engineering, tensile test is used to determine some factors of materials that are going to be tested such as elastic modulus, yield strength, ultimate strength etc.

In this test a tension exerted on a specimen that is made out of a particular material until it breaks and the instrument used to do the test gives a graph as output in which stress plotted against strain.

### $Stress(\sigma) =\frac{Force(F)}{Area(A)}$

### $Strain(\epsilon)=\frac{Change\space in\space Length(\delta)}{Original\space Length(L_o)}$

The curve given by the graph earlier mentioned in above paragraph can be divided into several regions. There is a region in which tensile curve act like almost straight line( following $y=mx$equation) and it is called linear region. To determine the yield point of the material, a line is drawn with a slope similar to that of the elastic modulus, but offset by 0.2% strain in the positive x-direction until it intersects the tensile curve. This intersection point is considered to be the yield point of the material, which acts as a boundary between the elastic and plastic regions.

If the material is ductile, after the plastic region, a phenomenon called necking occurs, which leads to the ultimate strength point and, finally, fracture.

If the material is brittle, it may not exhibit plastic behaviour or show plastic behaviour to a limited extent before fracture occurs.

# Procedure

```bash

# ------------------------ INITIALIZATION ----------------------------
units 		metal
dimension	3
boundary	p	p	p
atom_style	atomic
variable latparam equal 4.05

# ----------------------- ATOM DEFINITION ----------------------------
lattice		fcc ${latparam}
region		whole block 0 10 0 10 0 10
create_box	1 whole
lattice 	fcc ${latparam} orient x 1 0 0 orient y 0 1 0 orient z 0 0 1
create_atoms	1 region whole

# ------------------------ FORCE FIELDS ------------------------------
pair_style	eam/alloy
pair_coeff	* * Al99.eam.alloy Al

# ------------------------- SETTINGS ---------------------------------
compute csym all centro/atom fcc
compute peratom all pe/atom 

######################################
# EQUILIBRATION
reset_timestep	0
timestep 0.001
velocity all create 300 12345 mom yes rot no
fix 1 all npt temp 300 300 1 iso 0 0 1 drag 1 

# Set thermo output
thermo 1000
thermo_style custom step lx ly lz press pxx pyy pzz pe temp

# Run for at least 10 picosecond (assuming 1 fs timestep)
run 20000
unfix 1

# Store final cell length for strain calculations
variable tmp equal "lx"
variable L0 equal ${tmp}
print "Initial Length, L0: ${L0}"

######################################
# DEFORMATION
reset_timestep	0

fix 1 all npt temp 300 300 1 y 0 0 1 z 0 0 1 drag 1
variable srate equal 1.0e10
variable srate1 equal "v_srate / 1.0e12"
fix 2 all deform 1 x erate ${srate1} units box remap x

# Output strain and stress info to file
# for units metal, pressure is in [bars] = 100 [kPa] = 1/10000 [GPa]
# p2, p3, p4 are in GPa
variable strain equal "(lx - v_L0)/v_L0"
variable p1 equal "v_strain"
variable p2 equal "-pxx/10000"
variable p3 equal "-pyy/10000"
variable p4 equal "-pzz/10000"
fix def1 all print 100 " ${p1} ${p2} " file Al_SC_100.def1.txt screen no

# Use cfg for AtomEye
dump 		1 all custom 250 dump1.dump id type xs ys zs c_csym c_peratom fx fy fz vx vy vz
dump_modify 1 element Al

# Display thermo
thermo 	1000
thermo_style	custom step v_strain temp v_p2 v_p3 v_p4 ke pe press

run		50000

######################################
# SIMULATION DONE
print "All done"
```

when it comes to 600K,900K, and 1200k, all 300K temperature values in the input script were replaced with the relevant temperature.
