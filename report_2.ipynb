### Shell script for al outputs shown above

```bash
#!/bin/bash
for latconst in `seq 4.0 0.0025 4.1`;do 

cat <<EOF > m.txt
# ---------- Initialize Simulation --------------------- 
clear 
units metal 
dimension 3 
boundary p p p 
atom_style atomic 
atom_modify map array

# ---------- Create Atoms --------------------- 
lattice 	fcc ${latconst}
region	box block 0 1 0 1 0 1 units lattice
create_box	1 box

lattice	fcc ${latconst} orient x 1 0 0 orient y 0 1 0 orient z 0 0 1  
create_atoms 1 box
replicate 1 1 1

# ---------- Define Interatomic Potential --------------------- 
pair_style eam/alloy 
pair_coeff * * Al99.eam.alloy Al
neighbor 2.0 bin 
neigh_modify delay 10 check yes 
 
# ---------- Define Settings --------------------- 
compute eng all pe/atom 
compute eatoms all reduce sum c_eng 

variable natoms equal "count(all)" 
variable teng equal "c_eatoms"
variable length equal "lx"
variable ecoh equal "v_teng/v_natoms"

# ---------- Run Minimization --------------------- 
reset_timestep 0 
#fix 1 all box/relax iso 0.0 vmax 0.001
thermo 10 
thermo_style custom step pe press  c_eatoms v_ecoh etotal 
min_style cg 
minimize 1e-25 1e-25 5000 10000  

EOF

lmp < m.txt |tee m.out
 
CH=`grep Loop -B1 m.out|head -1|awk '{print $5}'`
P=`grep Loop -B1 m.out|head -1|awk '{print $3}'`
EA=`grep Loop -B1 m.out|head -1|awk '{print $4}'`

echo $latconst $EA >> eatoms.txt
echo $latconst $P >> press.txt
echo $latconst $CH >> cohesive.txt

done
```

# Discussion

After running the above shell script, obtain lattice parameter of the Aluminium was 4.05 while the experimental data for Aluminium was 4.0462. By rounding up the experimental data which is 4.0462 to two significant figures, gives a value of 4.05 Å, which is exactly same to the value obtained from the shell script.

In the second case, for the comparison of cohesive energy, it is necessary to convert the units of the Lammps software result into a format compatible with experimental data, which uses the unit of $\frac{kJ}{mol}$,whereas in the Lammps cohesive energy is calculated in $\frac{ev}{atm}$. As an starting step, first it needs to convert $eV \over atom$  to $kJ \over mole$ .

### $=\frac {-322.4\space kJ}{1\space mole}\space X \space \frac{1\space mole}{6.02214 X 10^{23}\space atom}\space  X  \space \frac{6.2415X10^{21}\space eV}{1\space kJ}$

$=-3.34\space \frac{eV}{atom}$

Now the two results are compatible in terms of units and evaluation can be carried out. Cohesive energy given by the Lammps software was $-3.36\space \frac{eV}{atom}$ where as in the experiment it was $-3.34\space \frac{eV}{atom}$

The two values slightly differ from each other. Differences in theories, methods, and conditions employed in the LAMMPS and in the experiments may account for this phenomenon.
