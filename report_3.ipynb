# Theories/Equations

### Melting

Let's explore what happens at the atomic level when we introduce heat to a material and continuously increase that heat beyond the material's melting point. Velocity is introduced using the 'velocity' command with a specified temperature. Additionally, we employ relevant ensembles, such as NVT, NPT, NVE, and more. In these ensembles, N, V, T, E, and P stand for the number of moles, volume, temperature, energy, and pressure, respectively. By utilising the LAMMPS command 'fix,' we can set up these ensembles with starting and ending temperature values.
Once the temperature of each atom increases, their kinetic energy rises, and they start to vibrate around their equilibrium positions. As the system's temperature continues to rise, there is a critical temperature at which the thermal energy overcomes the cohesive forces between atoms, causing the material to start melting. This temperature is known as the critical temperature.

### Ensembles

**Isothermal Isobaric Ensemble (NPT)**

Here, number of moles, pressure and temperature in the system are constant. Heat is allowed to transfer across the system boundaries. 

# Input Script

NVT ensemble

```bash
clear
units metal
dimension 3
atom_style atomic
atom_modify map array

timestep 0.005

lattice fcc 3.610 origin 0.0 0.0 0.0 orient x 1 0 0 orient y 0 1 0 orient z 0 0 1

region box block 0 5 0 5 0 5 units lattice
create_box 1 box
create_atoms 1 box

pair_style eam/alloy
pair_coeff * * Cu_Zhou04.eam.alloy Cu

# Equilibration

reset_timestep 0
velocity all create 300 12345 mom yes rot no dist gaussian
fix equalibration all npt temp 300 300 5 iso 1 1 5 drag 0.2
#fix data_e all print 1000  "step pxx pyy pzz lx ly lz vol temp etotal" file Cu.eq.txt screen no 

thermo 1000
thermo_style custom step pxx pyy pzz lx ly lz temp etotal ke

#Run
run 100000

# final volume
variable tmp equal "vol"
variable Vo equal ${tmp}
print "Volume initial is, Vo: ${Vo}"

#reset

unfix equalibration
#unfix data_e

#melting

reset_timestep 0

fix melt all npt temp 300 2000 5 iso 1 1 5 drag 0.2
#fix data_e all print 1000  "pxx pyy pzz lx ly lz vol temp etotal" file Cu_2000.txt screen no 
#fix data_m all print 1000 temp
variable eq8 equal "temp"
variable eq9 equal "vol/v_Vo"
fix data_m all print 1000 "${eq8} ${eq9}"  file melting.txt screen no
thermo 1000
thermo_style custom step pxx pyy pzz lx ly lz temp etotal ke

dump dump_1 all custom 100 Cu_npt.dump id type x y z ix iy iz vx vy vz 

run 100000

unfix melt
unfix data_m
undump dump_1
clear
```
