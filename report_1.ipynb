## Leonard Jones potential

This potential is named after Sir John Edward Lennard-Jones, mathematician who introduced the formula and used to calculate the potential energy of two non bonding atoms or molecules at different separation distances.

When two atoms are really close to each other (some times their electron clouds can be overlapped each other), repulsive force is dominantly effective and almost there is no attractive force acting on them( varies on the separation of atoms).

![Screenshot 2023-06-26 at 04.18.33.png](md_r1/Screenshot_2023-06-26_at_04.18.33.png)

But as the separation distance is keep getting increased, effect of repulsion force also get reduced and attraction force comes to the picture.

At some distance, where repulsion force is equal to the attraction force in magnitude the two forces going to be cancelled out and net potential energy between two atoms will be zero.(this particular distance is called as equilibrium distance and for reference this distance is known as $r$ in the report here onwards)

![Screenshot 2023-06-26 at 04.18.03.png](md_r1/Screenshot_2023-06-26_at_04.18.03.png)

As it is keep increasing the distance $r$ attraction force is more prominent than repulsion force as a result potential energy between two atoms again get increased, but this time it towards the negative Y direction. At some point of distance potential energy will occupy its maximum magnitude after this point if it is further separated two atoms attraction force will start to reduce and comes closer to zero potential energy at some distance of separation and start to act like a asymptote to the y=0 line and it will never meet zero condition regardless of how much the separation distance would be.

![Screenshot 2023-06-26 at 04.18.26.png](md_r1/Screenshot_2023-06-26_at_04.18.26.png)

 this relationship can be expressed mathematically as follows,

## $V(r)=4\epsilon[(\frac{\sigma}{r})^{12}-(\frac{\sigma}{r})^6]$

$V(r)=$ Intermolecular potential between the two atoms or molecules.

$\epsilon=$ Well depth and a measure of how strongly the two particles attract each other

$\sigma =$ distance at which the intermolecular potential between the two particles is zero

$r=$  distance of separation between both particles (measured from the centre of one particle to the centre of the other particle).

# Procedure

1. Lammps input script
    - write Lammps input script
    - initialise simulation
    - create atoms
    - define interatomic potential
    - set all the parameters that are needed in order to run the program
    
2. Run the Lammps input script and see the output

```bash
# Aluminium
# initialisation

clear
units metal
dimension 3
boundary p p p
atom_style atomic

# create atoms
lattice fcc 4.0479
region box block 0 3 0 3 0 3 units lattice 
create_box 1 box
create_atoms 1 region box

# atomic properties
mass 1 26.981539

# define interatomic potential
pair_style lj/cut 15
pair_coeff 1 1  0.2551 0.4080 #sigma epsilon

# run the programme
thermo 10
thermo_style custom step pe ke etotal 
timestep 0.001
dump dump_1 all custom 10 al.dump id type x y z ix iy iz vx vy vz 
run 100
undump dump_1
```

```bash
# Iron
# initialisation

clear
units metal
dimension 3
boundary p p p
atom_style atomic

# create atoms
lattice bcc 2.856
region box block 0 2 0 2 0 2 units lattice 
create_box 1 box
create_atoms 1 region box

# atomic properties
mass 1 55.845

# define interatomic potential
pair_style lj/cut 15
pair_coeff 1 1  0.4007 2.3193 

# run the programme
thermo 10
thermo_style custom step pe 
timestep 0.01
dump dump_1 all custom 10 fe.dump id type x y z ix iy iz vx vy vz 
run 100
undump dump_1
```

# Discussion
