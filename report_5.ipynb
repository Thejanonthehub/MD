Draw EV curve for aluminium using qtgrace

# Theories/Equations

EV curve of aluminium indicates the equilibrium position between aluminium atoms.

### Shell Script

```bash
#!/bin/bash
for a in `seq 3.9 0.025 4.2`;do

cat <<EOF > lmp.in
# ---------- Initialize Simulation --------------------- 
clear 
units metal 
dimension 3 
boundary p p p 
atom_style atomic 
atom_modify map array

# ---------- Create Atoms --------------------- 
lattice 	fcc ${a}
region	box block 0 1 0 1 0 1 units lattice
create_box	1 box

lattice	fcc ${a} orient x 1 0 0 orient y 0 1 0 orient z 0 0 1  
create_atoms 1 box
replicate 1 1 1

# ---------- Define Interatomic Potential --------------------- 
pair_style eam/alloy 
pair_coeff * * Al99.eam.alloy Al
neighbor 2.0 bin 
neigh_modify delay 10 check yes 
 
# ---------- Run 0 --------------------- 
run 0

EOF

lmp < lmp.in > lmp.out

E=`grep Loop -B1 lmp.out|head -1|awk '{print $5}'`

echo $a $E >> ev.txt

done
```

### Input Script

```bash
# ---------- Initialize Simulation --------------------- 
clear 
units metal 
dimension 3 
boundary p p p 
atom_style atomic 
atom_modify map array

# ---------- Create Atoms --------------------- 
lattice 	fcc 4.200
region	box block 0 1 0 1 0 1 units lattice
create_box	1 box

lattice	fcc 4.200 orient x 1 0 0 orient y 0 1 0 orient z 0 0 1  
create_atoms 1 box
replicate 1 1 1

# ---------- Define Interatomic Potential --------------------- 
pair_style eam/alloy 
pair_coeff * * Al99.eam.alloy Al
neighbor 2.0 bin 
neigh_modify delay 10 check yes 
 
# ---------- Run 0 --------------------- 
run 0
```
