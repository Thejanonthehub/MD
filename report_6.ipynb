### Finite Element Method (FEM)

FEM or Finite Element Method is a numerical techniques was developed in 1960s and solutions given by FEM method relies on problem data and the “mesh” of the geometric domain. Mesh consists of nodes and elements. 

# Procedure in brief

### Tensile test

1. First 3d specimen was created for the tensile test with aid of part module.
2. Switched to the mesh module and finite element mesh was generated for the specimen created(element type, element size) .
3. Material properties for the created specimen was set such as young’s modulus.
4. Suitable boundary conditions for tensile test setup were applied.
5. nodes in one face was fixed (z) while displacement boundary conditions were applied to nodes in opposite face (-z). 
6. Desired output parameters were set such as stress, strain.
7. finally though the submit function under job module following results were obtained.

### Heat Transfer

1. First 3d specimen was created for the tensile test with aid of part module.
2. Switched to the mesh module and finite element mesh was generated for the specimen created(element type, element size).
3. Next, material properties for the created specimen was set such as thermal, electrical, mechanical.
4. Interactions between thermal and mechanical analysis were defined.
5. Boundary conditions for the surface that supposed to be exposed to the head is going to applied were applied accordingly ( conduction, convection or radiation) 
6. Other boundary conditions were also defined such as symmetry
7. Then a stress was applied on one side of the specimen by mean of applying a displacement boundary condition.
8. finally though the submit function under job module following results were obtained.
